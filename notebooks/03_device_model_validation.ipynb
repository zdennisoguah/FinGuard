```python
import pandas as pd

customers = pd.read_csv(
    "../data/synthetic/customers.csv"
)

devices = pd.read_csv(
    "../data/synthetic/devices.csv"
)

customer_devices = pd.read_csv(
    "../data/synthetic/customer_devices.csv"
)
```

```python
devices.shape

customer_devices.shape
```

```python
devices["device_id"].is_unique
```

```python
customer_devices["customer_id"].nunique()
```

```python
customer_devices["device_id"].nunique()
```

```python
invalid_devices = customer_devices[
    ~customer_devices["device_id"].isin(
        devices["device_id"]
    )
]

len(invalid_devices)
```

```python
invalid_customers = customer_devices[
    ~customer_devices["customer_id"].isin(
        customers["customer_id"]
    )
]
len(invalid_customers)
```

```python
customer_devices.groupby(
    "device_id"
).size().sort_values(
    ascending=False
).head(20)
```

```python
shared_device_counts = (
    customer_devices
    .groupby("device_id")["customer_id"]
    .nunique()
)

shared_device_counts[
    shared_device_counts > 1
].sort_values(
    ascending=False
).head(20)
```

```python
duplicate_links = customer_devices.duplicated(
    subset=["customer_id", "device_id"]
).sum()

duplicate_links
```
